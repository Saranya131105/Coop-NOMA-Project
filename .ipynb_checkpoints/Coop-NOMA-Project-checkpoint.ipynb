```python
# Step 1: Import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Step 2: Define parameters
snr_values = np.arange(0, 21, 2)  # SNR range in dB
outage_coop = []
outage_noncoop = []

# Step 3: Simulation (dummy formulas for now)
for snr in snr_values:
    outage_coop.append(np.exp(-snr/10))       # placeholder formula
    outage_noncoop.append(np.exp(-snr/5))     # placeholder formula

# Step 4: Save results to CSV
df = pd.DataFrame({
    "SNR(dB)": snr_values,
    "Outage_Coop": outage_coop,
    "Outage_NonCoop": outage_noncoop
})
df.to_csv("outage_results.csv", index=False)

# Step 5: Plot results
plt.plot(snr_values, outage_coop, label="Cooperative NOMA")
plt.plot(snr_values, outage_noncoop, label="Non-Cooperative NOMA")
plt.xlabel("SNR (dB)")
plt.ylabel("Outage Probability")
plt.legend()
plt.grid(True)
plt.savefig("plots/outage_curve.png")
plt.show()
```
